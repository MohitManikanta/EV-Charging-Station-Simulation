```python
# EV Charging Calculator

battery_capacity = 40
current_charge = 20
target_charge = 100

charger_power = 7
electricity_rate = 8

# Calculate energy required
energy_needed = battery_capacity * (target_charge - current_charge) / 100

# Calculate charging time (ideal estimate)
charging_time = energy_needed / charger_power

# Calculate electricity cost
charging_cost = energy_needed * electricity_rate

print("===== EV CHARGING REPORT =====")

print("Energy Required:", energy_needed, "kWh")
print("Charging Time:", round(charging_time, 2), "hours")
print("Charging Cost: Rs.", charging_cost)
```

```python
print("===== EV CHARGING STATION SIMULATION =====")

# Station details
slots = int(input("Enter available charging slots: "))
vehicles = int(input("Enter number of vehicles: "))

while slots <= 0:
    print("Invalid slots!")
    slots = int(input("Enter available charging slots: "))

while vehicles <= 0:
    print("Invalid vehicle count!")
    vehicles = int(input("Enter number of vehicles: "))

total_energy = 0
total_cost = 0
total_time = 0

charging_now = min(slots, vehicles)
waiting = vehicles - charging_now

print("\nVehicles charging now:", charging_now)
print("Vehicles waiting:", waiting)

# Vehicle calculations
for i in range(vehicles):

    print("\n===== VEHICLE", i + 1, "=====")

    battery = float(input("Battery capacity (kWh): "))
    while battery <= 0:
        print("Battery must be positive!")
        battery = float(input("Battery capacity (kWh): "))

    current = float(input("Current charge (%): "))
    while current < 0 or current > 100:
        print("Enter charge between 0 and 100!")
        current = float(input("Current charge (%): "))

    target = float(input("Target charge (%): "))
    while target <= current or target > 100:
        print("Target must be above current and at most 100!")
        target = float(input("Target charge (%): "))

    charger = float(input("Charger power (kW): "))
    while charger <= 0:
        print("Charger power must be positive!")
        charger = float(input("Charger power (kW): "))

    rate = float(input("Electricity rate (Rs/kWh): "))
    while rate < 0:
        print("Rate cannot be negative!")
        rate = float(input("Electricity rate (Rs/kWh): "))

    energy = battery * (target - current) / 100
    time = energy / charger
    cost = energy * rate

    total_energy += energy
    total_cost += cost
    total_time += time

    print("\nEnergy required:", round(energy, 2), "kWh")
    print("Charging time:", round(time, 2), "hours")
    print("Charging cost: Rs.", round(cost, 2))

# Final report
print("\n===== FINAL STATION REPORT =====")
print("Total vehicles:", vehicles)
print("Vehicles charging now:", charging_now)
print("Vehicles waiting:", waiting)
print("Total energy:", round(total_energy, 2), "kWh")
print("Total charging cost: Rs.", round(total_cost, 2))
print("Sum of individual charging times:",
      round(total_time, 2), "hours")

print("\n===== SIMULATION COMPLETED =====")
```
